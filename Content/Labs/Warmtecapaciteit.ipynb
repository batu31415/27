# Bepaling van warmtecapaciteit van een onbekend materiaal


# Introductie

Onbekende materialen kunnen geïdentificeerd worden door hun eigenschappen te meten. Een van deze eigenschappen is de warmtecapaciteit. In dit practicum gaan we de warmtecapaciteit van een onbekend materiaal bepalen door middel van een calorimeter experiment. Daarbij wordt een bepaalde massa van het materiaal naar een bekende temperatuur gebracht waarna het in een bekende hoeveelheid water met bekende temperatuur wordt geplaatst. Door de temperatuur van het water te meten na het mengen kan de warmtecapaciteit van het onbekende materiaal worden berekend.

# Theorie

De soortelijke warmte $c$ van een materiaal is gedefinieerd als de hoeveelheid warmte $Q$ die nodig is om de temperatuur $T$ van een kilogram van het materiaal met één graad Celsius (of één Kelvin) te verhogen:

$$
    c = \frac{Q}{m \Delta T}
$$ (eq:heat_capacity)

Waarbij $Q$ de hoeveelheid warmte in Joules is, $m$ de massa in kilogram is en $\Delta T$ de verandering in temperatuur is. Gegeven de wet van Black, die stelt dat de totale hoeveelheid warmte in een geïsoleerd systeem constant blijft, kunnen we de warmte die het onbekende materiaal verliest gelijkstellen aan de warmte die het water opneemt:

$$
    Q_{materiaal} = -Q_{water} 
$$ (eq:black)

wanneer we de massa's en de begintemperaturen van beide systemen kennen, maar slechts een van de twee soortelijke warmtes, kunnen we de onbekende soortelijke warmte berekenen. We combineren vergelijkingen [](#eq:heat_capacity) en [](#eq:black) om de volgende vergelijking te krijgen:

$$
    T_e = \frac{c_w m_w T_{w,b}+c_m m_m T_{m,b}}{c_w m_w + c_m m_m}
$$ (eq:combined)

Waarbij de subscripts $b$ en $e$ respectievelijk staan voor begintoestand en eindtoestand, $w$ voor water en $m$ voor het onbekende materiaal.

Bij metingen aan verschillende massa's van het onbekende materiaal en vervolgens een least square fit aan bovenstaande vergelijking kunnen we een precieze waarde voor de soortelijke warmte van het onbekende materiaal bepalen. Dat is, wanneer de warmtecapaciteit van bijvoorbeeld de beker te verwaarlozen is.

# Resultaten


````python
import numpy as np
import matplotlib.pyplot as plt

# Constante
c_w = 4186  # J/(kg·K)

# Data per meting
metingen = [
    (284.9 - 145.7, 51.9, 20.2, 69.9, 21.6),  # meting 1
    (245.2 - 145.7, 26.1, 20.2, 69.9, 21.2),  # meting 2
    (237.8 - 145.7, 26.0, 20.6, 69.9, 21.8),  # meting 3
    (263.9 - 145.7, 13.1, 20.0, 69.9, 20.6),  # meting 4
]

# Bereken c_m per meting
c_m = []
for m_w_g, m_m_g, T_wb, T_mb, T_e in metingen:
    m_w = m_w_g / 1000  # g -> kg
    m_m = m_m_g / 1000  # g -> kg
    c = (c_w * m_w * (T_e - T_wb)) / (m_m * (T_mb - T_e))
    c_m.append(c)

# Gemiddelde
c_avg = sum(c_m) / len(c_m)

# Plot
x = [1, 2, 3, 4]
plt.plot(x, c_m, "o-", label="metingen")
plt.axhline(c_avg, linestyle="--", label=f"gemiddelde = {c_avg:.0f}")

plt.xlabel("Meting")
plt.ylabel("c_m (J/(kg·K))")
plt.title("Soortelijke warmte per meting")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
````

## Discussie en conclusie

Uit de vier metingen volgt voor de soortelijke warmte van het onbekende materiaal ongeveer:

- $c_{m,1}\approx 325~\mathrm{J/(kg\cdot K)}$
- $c_{m,2}\approx 328~\mathrm{J/(kg\cdot K)}$
- $c_{m,3}\approx 370~\mathrm{J/(kg\cdot K)}$
- $c_{m,4}\approx 460~\mathrm{J/(kg\cdot K)}$

Het gemiddelde over alle metingen is:

$$
\overline{c_m}\approx 371~\mathrm{J/(kg\cdot K)},
$$

met een relatief grote spreiding (ongeveer $\sigma \approx 63~\mathrm{J/(kg\cdot K)}$).

In de grafiek is te zien dat meting 1 en 2 goed overeenkomen, meting 3 iets hoger uitkomt en meting 4 duidelijk afwijkt (hogere waarde).
Deze spreiding is te verklaren doordat het experiment uitgaat van een ideale warmtebalans:

$$
c_w\,m_w\,(T_e - T_{w,b}) = c_m\,m_m\,(T_{m,b} - T_e).
$$

Hierbij wordt aangenomen dat er geen warmte verloren gaat aan het bekertje, de thermometer of de omgeving, en dat het blokje niet merkbaar afkoelt tijdens het verplaatsen.
In werkelijkheid treden warmteverliezen op en kan het blokje al (gedeeltelijk) afkoelen voordat het in het water komt.

Meting 4 is extra gevoelig voor fouten omdat de massa van het blokje daar klein is.
Bij een kleine $m_m$ leidt een kleine meetfout in temperatuur (of extra warmteverlies) tot een relatief grote verandering in de berekende $c_m$.
Ook onvoldoende mengen, een andere mengtijd, of water dat aan het blokje blijft hangen kan het resultaat beïnvloeden.

## Conclusie en implicaties

Op basis van het gemiddelde volgt:

$$
c_m \approx 3.7 \times 10^2~\mathrm{J/(kg\cdot K)}.
$$

Dit wijst op een materiaal met relatief lage warmtecapaciteit, wat past bij een metaal in de orde van grootte van koper- of messingachtige materialen (veel lager dan aluminium).

De spreiding in de metingen (met name de hoge waarde van meting 4) laat zien dat het resultaat sterk beïnvloed wordt door warmteverlies en meetonnauwkeurigheid, vooral bij kleine massa’s.
Het experiment maakt het daarom mogelijk om het materiaal globaal te classificeren, maar een preciezere identificatie vereist verbeteringen zoals betere isolatie (bijv. deksel), correctie voor de warmtecapaciteit van beker en thermometer, sneller werken en beter mengen, en/of werken met grotere massa’s of grotere temperatuurverschillen.
