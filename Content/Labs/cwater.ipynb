# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

Water warmt op wanneer er energie aan wordt toegevoegd: die energie vergroot de inwendige energie van het water en daardoor stijgt de temperatuur. De hoeveelheid energie Q die nodig is om een massa m een temperatuurverandering $\Delta$ T te geven wordt beschreven door: Q = mc $\Delta$ T, waarbij c de soortelijke warmte is. 
In dit experiment wordt het water verwarmd met een elektrisch element. Daarvoor geldt dat de vermogen P = I^2R is. De toegevoegde energie is E = pt. Als deze energie in het water terechtkomt, geldt dat: P = mc dT/dt.
dT/dt kan door de grafiek bepaald worden, en om deze manier kan de soortelijke warmte van water bepaald worden.

## Data analyse
De gemeten temperatuur T is uitgezet tegen de tijd t. Daarbij is gebruik gemaakt van een tijdsinterval van 30 seconden. Er is een linear fit gebruikt om de helling te bepalen van de metingen. De verkregen helling is vervolgens gebruikt voor de uiteindelijke berekening van de soortelijke warmte.

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

P = 19.2 #W (gemeten)
t = np.array([0,30, 60, 90, 120, 180, 210, 240, 270, 300, 330, 360,
              390, 420, 450, 480, 510, 540, 570, 600, 630])

T = np.array([21.4, 22.2, 23.8, 24.6, 25.5, 26.4, 27.2, 28.1, 29.4, 30, 30.9, 31.9, 32.8, 33.5, 34.3, 35, 35.8, 36.6, 37.4, 38.2, 39])

def f(x, a, b):
    return a*x + b

var, cov = curve_fit(f, t, T)
a, b = var
a_unc, b_unc = np.sqrt(np.diag(cov))

print(f"helling a = {a:.6f} +/- {a_unc:.6f} C/s")
print(f"in C/min = {a*60:.3f} +/- {a_unc*60:.3f} C/min")

plt.plot(t,T)
plt.plot(t, f(t, a, b))
plt.xlabel('tijd')
plt.ylabel('Temperatuur')
plt.show() 



# Sla figuren op met  
# 
# plt.savefig("figures/naam.png", dpi=450)

````

# Discussie en conclusie

De temperatuur stijgt vrijwel lineair met de tijd, wat past bij verwarming met ongeveer constant elektrisch vermogen. Uit de lineire fit volgt dat dT/dt = 0.02759 K/s. Met P = 19.2 W en een massa van 0.165 kg volgt dat c = P/(m(dT/dt)) = 4217 +/- 46 J/(kg * K), wat redelijk dichtbij de literatuurwaarde ligt. Afwijken komen door systematische fouten zoals warmteverlies aan de omgeving en het feit dat het verwarmingselement ook opwarmt. 

Ter conclusie kan er worden gezegd dat bij constant vermogen, het water linear opwarmt. Hieruit volgt dat de soortelijk warmte 4217+/- 46J/(kg * K) is, wat bijna overeenkomt met de literatuurwaarde.
